# 📈 Monthly Sales Trend Analysis | Python Data Analytics

```python
import pandas as pd
import matplotlib.pyplot as plt

gsd['Order Date'] = pd.to_datetime(gsd['Order.Date'])

monthly_sales = gsd.groupby(
    gsd['Order Date'].dt.to_period('M')
)['Sales'].sum()

monthly_sales.index = monthly_sales.index.to_timestamp()

plt.figure(figsize=(14,5))
plt.plot(monthly_sales.index, monthly_sales.values,
         marker='o', linewidth=2)

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.grid(True)
plt.show()
```

## 🔍 Project Overview

Analyze monthly sales performance using Python, Pandas, and Matplotlib to identify sales trends, seasonal patterns, peak-performing periods, and potential business growth opportunities.

## 🛠️ Tools & Technologies

- 🐍 **Python** — Data analysis and processing
- 🐼 **Pandas** — Date conversion, grouping, and aggregation
- 📊 **Matplotlib** — Monthly sales trend visualization

## 💡 Key Business Insights

- 📈 **Sales Trend Monitoring:** Track monthly sales fluctuations to understand overall business performance.

- 🚀 **Peak Sales Identification:** Identify the months with the highest sales to understand when revenue performance is strongest.

- 📉 **Sales Decline Detection:** Highlight periods of falling sales that may require further investigation.

- 🔄 **Seasonality Analysis:** Compare recurring monthly patterns across different years to identify potential seasonal demand.

- 🎯 **Business Growth Opportunities:** Use monthly sales patterns to support inventory planning, promotional campaigns, and revenue forecasting.

## 🎯 Business Objective

Transform historical sales data into actionable insights that support data-driven decisions, improve sales planning, and strengthen business performance.

## 📌 Conclusion

Monthly sales trend analysis provides a clear overview of revenue performance over time. By identifying peak periods, declining trends, and recurring patterns, businesses can make more informed strategic decisions.

# 📊 Sales vs Profit Analysis | Python Data Analytics

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))

sns.scatterplot(
    data=gsd,
    x='Sales',
    y='Profit',
    hue='Category',
    alpha=0.6
)

plt.title('Sales vs Profit')
plt.show()
```

- 🏆 **Category Performance:** Compare product categories to discover which categories demonstrate stronger profitability patterns.

- ⚠️ **Loss-Making Transactions:** Identify observations with negative profit and investigate potential causes, such as excessive discounts, high costs, or pricing issues.

- 🎯 **High-Value Transactions:** Examine transactions with high sales and high profit to understand opportunities for profitable growth.

# 🏆 Top 10 Sub-Categories by Sales | Python Data Analytics

```python
top_sub = (
    gsd.groupby('Sub.Category')['Sales']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(10,6))
top_sub.plot(kind='bar', color='steelblue')

plt.title('Top 10 Sub-Categories by Sales')
plt.ylabel('Sales')
plt.xticks(rotation=45)

plt.show()
```


## 🔍 Project Overview

- 📊 **Sales Performance Ranking:** Compare the top 10 sub-categories to evaluate their relative contribution to overall sales.

- 🔎 **Performance Gaps:** Examine the difference between leading and lower-ranked sub-categories to identify areas that may need additional analysis.

- 📦 **Inventory Planning:** Use sales performance patterns to help forecast demand and allocate inventory more effectively.

- 🎯 **Marketing Strategy:** Develop targeted campaigns around high-demand product segments while investigating opportunities to improve weaker performers.

- 💼 **Business Decision Support:** Use product-level sales rankings to inform assortment planning, resource allocation, and strategic sales decisions.

# 💰 Profit by Category Analysis | Python Data Analytics

```python
category_profit = (
    gsd.groupby('Category')['Profit']
    .sum()
    .sort_values()
)

plt.figure(figsize=(8,5))
category_profit.plot(kind='barh', color='green')

plt.title('Profit by Category')
plt.xlabel('Profit')
plt.show()
```

# 🌎 Region vs Category Sales Analysis | Python Data Analytics

```python
import seaborn as sns

pivot = pd.pivot_table(
    gsd,
    values='Sales',
    index='Region',
    columns='Category',
    aggfunc='sum'
)

plt.figure(figsize=(10,6))

sns.heatmap(
    pivot,
    annot=True,
    fmt='.0f',
    cmap='YlGnBu'
)

plt.title('Region vs Category Sales')
plt.show()
```

## 🔍 Project Overview
Analyze sales distribution across geographical regions and product categories using Python, Pandas, Seaborn, and Matplotlib. This heatmap highlights regional sales patterns and helps compare category-level revenue performance across different markets.

## 🛠️ Tools & Technologies
- 🐍 **Python** — Data analysis and processing
- 🐼 **Pandas** — Pivot tables and sales aggregation
- 🎨 **Seaborn** — Heatmap visualization
- 📊 **Matplotlib** — Chart formatting and presentation

## 💡 Key Business Insights

- 🌍 **Regional Sales Performance:** Compare total sales across regions to identify stronger and weaker revenue-generating markets.

- 🏆 **Category Leaders by Region:** Identify which product category records the highest sales within each region.

- 🔥 **High-Performing Markets:** Highlight region–category combinations with strong sales performance and investigate the factors driving demand.

- 📉 **Sales Performance Gaps:** Identify lower-sales combinations that may require further investigation into customer demand, product availability, or marketing effectiveness.

- 🎯 **Regional Marketing Strategy:** Use sales patterns to support region-specific promotions and targeted marketing campaigns.

- 📦 **Inventory Optimization:** Align inventory planning with regional category-level sales demand to help reduce stock shortages and excess inventory.

- 🚀 **Business Growth Opportunities:** Identify promising category–region combinations for expansion and additional sales initiatives.

## 🎯 Business Objective
Understand how sales vary across regions and product categories to support regional planning, inventory optimization, targeted marketing, and data-driven business decisions.

## 📌 Conclusion
The Region vs Category Sales heatmap provides a visual comparison of revenue across geographical markets and product segments. It helps decision-makers identify high-performing combinations, investigate sales gaps, and develop more effective regional business strategies.
